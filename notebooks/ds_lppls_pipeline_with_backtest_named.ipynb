# DS‑LPPLS — Pipeline hebdomadaire (fenêtres 13–104)

Entrée : un fichier CSV type `SBF120.csv` (séparateur `;`, décimales `,`, 3 lignes d’entête Start/End/Frequency).

Sorties :
- `details_by_scale_all_series.csv`
- `summary_by_scale_wide.csv`
- `bubble_summary_by_scale_wide.csv` (format **identique** à `summary_by_scale_wide.csv` + colonnes bulles)
- `bubble_type_by_asset.csv` (typologie court/long/cohérence + `priority_score`)
- `portfolio_construction.csv` (poids long-only et long/short)
- `heatmap_global.png`

⚠️ Ce notebook utilise un ajustement LPPLS **rapide** (grille (m, ω, tc) + moindres carrés linéaires) pour obtenir des signaux et horizons.


## 0) Configuration (à modifier à chaque run)


```python

from pathlib import Path

# --- Entrée ---
INPUT_CSV = Path("/content/SBF120.csv")  # <-- à modifier (ou déposer le fichier dans Colab)

# --- Sorties ---
OUTPUT_DIR = Path("/content/lppls_outputs")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# --- Fenêtres hebdomadaires ---
WINDOW_WEEKS = [13, 26, 39, 52, 65, 78, 104]

# --- Noms des fichiers ---
DETAILS_OUT = OUTPUT_DIR / "details_by_scale_all_series.csv"
SUMMARY_OUT = OUTPUT_DIR / "summary_by_scale_wide.csv"
BUBBLETYPE_OUT = OUTPUT_DIR / "bubble_type_by_asset.csv"
BUBBLE_WIDE_OUT = OUTPUT_DIR / "bubble_summary_by_scale_wide.csv"
PORTFOLIO_OUT = OUTPUT_DIR / "portfolio_construction.csv"
HEATMAP_OUT = OUTPUT_DIR / "heatmap_global.png"

print("INPUT_CSV:", INPUT_CSV)
print("OUTPUT_DIR:", OUTPUT_DIR)
```

## 1) Chargement et nettoyage (CSV `;` + décimales `,`)


```python

import pandas as pd
import numpy as np

# Le fichier contient 3 lignes d'entête (Start/End/Frequency). La 4e ligne est l'en-tête des colonnes.
df = pd.read_csv(INPUT_CSV, sep=";", skiprows=3, header=0, encoding="utf-8").rename(columns={"Name":"date"})
df["date"] = pd.to_datetime(df["date"], dayfirst=True, errors="coerce")
df = df.dropna(subset=["date"]).sort_values("date")

# Conversion virgule -> point puis float
for c in df.columns[1:]:
    df[c] = pd.to_numeric(df[c].astype(str).str.replace(",", ".", regex=False), errors="coerce")

df = df.set_index("date")

print("Shape (daily):", df.shape)
df.head(3)
```

## 2) Passage en hebdomadaire (dernier cours de la semaine)


```python

weekly = df.resample("W-FRI").last().dropna(how="all")
print("Shape (weekly):", weekly.shape)
weekly.tail(3)
```

## 3) Ajustement LPPLS rapide (grille)


```python

import numpy as np

def fast_lppls_fit(logp: np.ndarray):
    """
    Ajustement LPPLS rapide :
    - grille sur m, omega (ω), tc
    - estimation linéaire de A, B, C1, C2
    Retourne le meilleur fit (SSE minimum).
    """
    y = np.asarray(logp, float)
    n = len(y)
    t = np.arange(n, dtype=float)

    # Grilles (compactes pour rester rapides)
    m_grid = np.linspace(0.1, 0.9, 7)
    omega_grid = np.linspace(6.0, 13.0, 7)
    ahead_max = min(1.5*n, max(6, n))
    tc_ahead_grid = np.linspace(1.0, ahead_max, 7)

    best = None
    for m in m_grid:
        for omega in omega_grid:
            for ahead in tc_ahead_grid:
                tc = (n - 1) + ahead
                tau = tc - t
                if np.any(tau <= 0):
                    continue

                x1 = tau**m
                ang = omega*np.log(tau)
                x2 = x1*np.cos(ang)
                x3 = x1*np.sin(ang)
                X = np.column_stack([np.ones(n), x1, x2, x3])

                beta, *_ = np.linalg.lstsq(X, y, rcond=None)
                yhat = X @ beta
                resid = yhat - y
                sse = float(np.sum(resid**2))

                if (best is None) or (sse < best["sse"]):
                    A, B, C1, C2 = beta
                    sst = float(np.sum((y - y.mean())**2))
                    r2 = 1 - sse/sst if sst > 0 else np.nan
                    rmse = float(np.sqrt(np.mean(resid**2)))
                    C = float(np.sqrt(C1*C1 + C2*C2))
                    phi = float(np.arctan2(C2, C1) % (2*np.pi))
                    best = dict(success=True, A=float(A), B=float(B), C=C, m=float(m),
                                omega=float(omega), phi=phi, tc=float(tc),
                                r2=r2, rmse=rmse, sse=sse, n=n)
    return best if best is not None else {"success": False, "n": len(y)}

def classify_fit(fit: dict):
    """Filtrage minimal + signal UP/DOWN."""
    if not fit.get("success", False):
        return False, "NONE"
    n = fit["n"]
    m, omega, tc, r2 = fit["m"], fit["omega"], fit["tc"], fit["r2"]

    valid = True
    if not (0.1 <= m <= 0.9): valid = False
    if not (6.0 <= omega <= 13.0): valid = False
    if not ((n-1) < tc <= (n-1) + 1.5*n): valid = False
    if np.isnan(r2) or r2 < 0.2: valid = False

    if not valid:
        return False, "NONE"

    # Convention : B < 0 => bulle haussière (accélération super‑exponentielle)
    signal = "UP" if fit["B"] < 0 else "DOWN"
    return True, signal
```

## 4) Calcul par action et par fenêtre (fin de période) → `details_by_scale_all_series.csv`


```python

details_rows = []
for asset in weekly.columns:
    s = weekly[asset].dropna()
    if len(s) < max(WINDOW_WEEKS):
        continue

    for w in WINDOW_WEEKS:
        window = s.iloc[-w:]
        if len(window) < w:
            continue

        y = np.log(window.values.astype(float))
        fit = fast_lppls_fit(y)
        is_valid, signal = classify_fit(fit)
        tc_weeks_ahead = (fit["tc"] - (fit["n"]-1)) if fit.get("success", False) else np.nan

        details_rows.append({
            "asset": asset,
            "window_weeks": w,
            "start_date": window.index.min().date().isoformat(),
            "end_date": window.index.max().date().isoformat(),
            "n_obs": fit.get("n", len(y)),
            "A": fit.get("A", np.nan),
            "B": fit.get("B", np.nan),
            "C": fit.get("C", np.nan),
            "m": fit.get("m", np.nan),
            "omega": fit.get("omega", np.nan),
            "phi": fit.get("phi", np.nan),
            "tc_weeks_ahead": tc_weeks_ahead,
            "r2": fit.get("r2", np.nan),
            "rmse": fit.get("rmse", np.nan),
            "is_valid": bool(is_valid),
            "signal": signal
        })

details = pd.DataFrame(details_rows)
details.to_csv(DETAILS_OUT, index=False)
print("Saved:", DETAILS_OUT, "| rows:", len(details), "| assets:", details["asset"].nunique())
details.head(3)
```

## 5) Synthèse wide (format `summary_by_scale_wide.csv`)


```python

def make_wide(metric: str) -> pd.DataFrame:
    tmp = details.pivot(index="asset", columns="window_weeks", values=metric)
    tmp.columns = [f"{metric}_{c}w" for c in tmp.columns]
    return tmp

wide = pd.concat([
    make_wide("signal"),
    make_wide("is_valid"),
    make_wide("tc_weeks_ahead"),
    make_wide("r2"),
], axis=1).reset_index()

valid_cols = [c for c in wide.columns if c.startswith("is_valid_")]
wide["valid_frac"] = wide[valid_cols].mean(axis=1)

# Médiane de tc sur fenêtres valides
tc_cols = {int(c.split("_")[2].replace("w","")): c for c in valid_cols}
tc_value_cols = {int(c.split("_")[3].replace("w","")): c for c in wide.columns if c.startswith("tc_weeks_ahead_")}
tc_medians = []
for _, row in wide.iterrows():
    tcs = []
    for w, vcol in tc_cols.items():
        if bool(row[vcol]):
            tcs.append(row[tc_value_cols[w]])
    tc_medians.append(float(np.nanmedian(tcs)) if len(tcs) else np.nan)

wide["tc_weeks_ahead_median"] = tc_medians

wide.to_csv(SUMMARY_OUT, index=False)
print("Saved:", SUMMARY_OUT, "| rows:", len(wide))
wide.head(3)
```

## 6) Livrable par type de bulle (court terme / long terme / multi‑échelles)


```python

short_ws = [13, 26]
long_ws = [78, 104]

def agg_signal(row, ws):
    sigs = []
    for w in ws:
        s = row.get(f"signal_{w}w", None)
        v = row.get(f"is_valid_{w}w", False)
        if bool(v) and isinstance(s, str) and s in ("UP", "DOWN"):
            sigs.append(s)
    if len(sigs) == 0:
        return "NONE"
    up = sum(1 for x in sigs if x == "UP")
    down = sum(1 for x in sigs if x == "DOWN")
    if up > down: return "UP"
    if down > up: return "DOWN"
    return "MIXED"

bubble_rows = []
for _, row in wide.iterrows():
    st = agg_signal(row, short_ws)
    lt = agg_signal(row, long_ws)

    if st in ("UP", "DOWN") and lt == st:
        mult = f"COHERENT_MULTI_{st}"
    elif st == "NONE" and lt == "NONE":
        mult = "NO_BUBBLE"
    else:
        mult = "MIXED_SCALES"

    tc_med = row["tc_weeks_ahead_median"]
    horizon_penalty = 1.0 if np.isnan(tc_med) else (1.0 + max(tc_med, 0)/10.0)
    coh_bonus = 2.0 if mult.startswith("COHERENT_MULTI_") else (0.5 if mult == "MIXED_SCALES" else 0.0)
    priority = 100.0 * (coh_bonus + row["valid_frac"]) / horizon_penalty

    bubble_rows.append({
        "asset": row["asset"],
        "bubble_short_term": st,
        "bubble_long_term": lt,
        "bubble_multiscale_state": mult,
        "valid_frac": row["valid_frac"],
        "tc_weeks_ahead_median": tc_med,
        "priority_score": priority
    })

bubble_type = pd.DataFrame(bubble_rows).sort_values("priority_score", ascending=False)
bubble_type.to_csv(BUBBLETYPE_OUT, index=False)
print("Saved:", BUBBLETYPE_OUT, "| rows:", len(bubble_type))
bubble_type.head(10)
```

## 7) Résultats des bulles au format wide (même structure que `summary_by_scale_wide.csv`)


```python

bubble_cols = ["bubble_short_term","bubble_long_term","bubble_multiscale_state","valid_frac","tc_weeks_ahead_median","priority_score"]
bubble_wide = wide.merge(bubble_type[["asset"] + bubble_cols], on="asset", how="left", suffixes=("", "_bubble"))

# wide contient déjà valid_frac et tc_weeks_ahead_median : on remplace par ceux du livrable bulles
for col in ["valid_frac", "tc_weeks_ahead_median"]:
    if col in wide.columns and col+"_bubble" in bubble_wide.columns:
        bubble_wide = bubble_wide.drop(columns=[col]).rename(columns={col+"_bubble": col})

bubble_wide.to_csv(BUBBLE_WIDE_OUT, index=False)
print("Saved:", BUBBLE_WIDE_OUT, "| rows:", len(bubble_wide))
bubble_wide.head(3)
```

## 8) Heatmap globale (UP=1, DOWN=-1, invalide=0)


```python

import matplotlib.pyplot as plt

sig_map = {"UP": 1, "DOWN": -1, "NONE": 0, "MIXED": 0}
heat = details.copy()
heat["signal_num"] = heat["signal"].map(sig_map).fillna(0) * heat["is_valid"].astype(int)

heat_pivot = heat.pivot(index="asset", columns="window_weeks", values="signal_num")
# tri par priority_score (optionnel) si présent
order = bubble_type["asset"].tolist()
heat_pivot = heat_pivot.reindex(order)

plt.figure(figsize=(10, max(6, len(heat_pivot)/8)))
plt.imshow(heat_pivot.values, aspect="auto", interpolation="nearest")
plt.yticks(range(len(heat_pivot.index)), heat_pivot.index, fontsize=6)
plt.xticks(range(len(heat_pivot.columns)), [f"{c}w" for c in heat_pivot.columns])
plt.title("Heatmap globale — signal LPPLS (UP=1, DOWN=-1, invalide=0)")
plt.colorbar()
plt.tight_layout()
plt.savefig(HEATMAP_OUT, dpi=200)
plt.close()

print("Saved:", HEATMAP_OUT)
```

## 9) Livrable portefeuille (long-only & long/short)


```python

# Volatilité hebdo (52 semaines) et inverse-vol pour pondération simple
rets = weekly.pct_change()
vol_52 = rets.tail(52).std(skipna=True)
inv_vol = (1 / vol_52.replace(0, np.nan)).replace([np.inf, -np.inf], np.nan)

bt = bubble_type.set_index("asset")

eligible_long = bt[bt["bubble_multiscale_state"] != "COHERENT_MULTI_UP"].index
w_long = inv_vol.reindex(eligible_long).fillna(inv_vol.median()).clip(lower=0)
w_long = w_long / w_long.sum()

short_assets = bt[bt["bubble_multiscale_state"] == "COHERENT_MULTI_UP"].index
w_short = bt.loc[short_assets, "priority_score"].clip(lower=0)
w_short = (w_short / w_short.sum()) if (len(w_short) > 0 and w_short.sum() > 0) else pd.Series(dtype=float)

w_long_ls = inv_vol.reindex(eligible_long).fillna(inv_vol.median()).clip(lower=0)
w_long_ls = (w_long_ls / w_long_ls.sum()) if w_long_ls.sum() > 0 else w_long_ls

gross_long = 1.0 if len(w_short) == 0 else 0.5
gross_short = 0.0 if len(w_short) == 0 else 0.5

def rationale(state):
    if state == "COHERENT_MULTI_UP":
        return "Sous-pondération / short (bulle haussière cohérente)"
    if state == "COHERENT_MULTI_DOWN":
        return "Surveillance (bulle baissière cohérente)"
    if state == "MIXED_SCALES":
        return "Signal mixte (prudence)"
    return "Pas de bulle multi-échelles cohérente"

portfolio = pd.DataFrame({"asset": bt.index})
portfolio["bubble_multiscale_state"] = bt["bubble_multiscale_state"].values
portfolio["priority_score"] = bt["priority_score"].values
portfolio["weight_long_only"] = portfolio["asset"].map(w_long).fillna(0.0).values
portfolio["weight_long_short"] = (
    portfolio["asset"].map(w_long_ls).fillna(0.0) * gross_long
    - portfolio["asset"].map(w_short).fillna(0.0) * gross_short
).values
portfolio["rationale"] = portfolio["bubble_multiscale_state"].map(rationale).values

portfolio.to_csv(PORTFOLIO_OUT, index=False)
print("Saved:", PORTFOLIO_OUT, "| rows:", len(portfolio))
portfolio.head(10)
```

## Backtest (version corrigée : noms d'actions)

Cette section ajoute un backtest hebdomadaire avec :
- données SBF 120 (fréquence hebdomadaire, clôture vendredi),
- fenêtres max 2 ans (104 semaines),
- rebalancement toutes les 4 semaines (holding 4 semaines),
- rebalancement **uniquement** si le statut change (lazy rebalance),
- exclusion des actions sans historique suffisant,
- deux stratégies : Long-only et Long/Short/Flat,
- génération des poids en format **long** et **wide** (dates en lignes, actions en colonnes),
- métriques : turnover, max drawdown, Sharpe (annualisé sur base hebdo),
- graphe NAV.

⚠️ Correction principale : le fichier `SBF120.csv` contient une ligne d'en-tête `Name;ABIVAX;...`.  
On lit donc avec `skiprows=3` (et non `skiprows=4`) pour conserver les **noms** d'actions.


```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =========================
# I/O
# =========================
INPUT_CSV = Path("SBF120.csv")  # même dossier que ce notebook
OUTDIR = Path("lppls_backtest_outputs_named")
OUTDIR.mkdir(parents=True, exist_ok=True)

# =========================
# Lecture SBF120.csv (format : ; + virgules décimales + ligne "Name" avec les tickers)
# =========================
df = pd.read_csv(INPUT_CSV, sep=";", skiprows=3, engine="python")
date_col = df.columns[0]  # colonne "Name" contenant les dates
df[date_col] = pd.to_datetime(df[date_col], dayfirst=True, errors="coerce")
df = df.dropna(subset=[date_col]).sort_values(date_col).set_index(date_col)

def to_float_df(dff: pd.DataFrame) -> pd.DataFrame:
    out = dff.copy()
    for c in out.columns:
        out[c] = pd.to_numeric(out[c].astype(str).str.replace(",", ".", regex=False), errors="coerce")
    return out

prices = to_float_df(df)

# Hebdomadaire (clôture vendredi)
weekly = prices.resample("W-FRI").last()
weekly_rets = weekly.pct_change(fill_method=None)

assets = list(weekly.columns)

# =========================
# Paramètres backtest
# =========================
WINDOW_WEEKS = [13, 26, 39, 52, 65, 78, 104]
start_date = pd.Timestamp("2024-02-01")
hold_weeks = 4
lookback_max = 104  # 2 ans

weekly_idx = weekly.index[weekly.index >= start_date]
rebalance_dates = list(weekly_idx[::hold_weeks])
rebalance_dates = [d for d in rebalance_dates if (weekly.index.get_loc(d) + hold_weeks) < len(weekly.index)]
if len(rebalance_dates) > 24:
    rebalance_dates = rebalance_dates[:24]

# =========================
# LPPLS rapide (grilles réduites : backtest)
# =========================
def fast_lppls_fit_tiny(logp: np.ndarray):
    y = np.asarray(logp, float)
    n = len(y)
    t = np.arange(n, dtype=float)

    m_grid = np.array([0.2, 0.5, 0.8])
    omega_grid = np.array([7.0, 9.5, 12.0])
    ahead_max = min(1.5 * n, max(6, n))
    tc_ahead_grid = np.linspace(1.0, ahead_max, 3)

    best = None
    for m in m_grid:
        for omega in omega_grid:
            for ahead in tc_ahead_grid:
                tc = (n - 1) + ahead
                tau = tc - t
                if np.any(tau <= 0):
                    continue
                x1 = tau ** m
                ang = omega * np.log(tau)
                X = np.column_stack([np.ones(n), x1, x1 * np.cos(ang), x1 * np.sin(ang)])
                beta, *_ = np.linalg.lstsq(X, y, rcond=None)
                resid = X @ beta - y
                sse = float(np.dot(resid, resid))
                if (best is None) or (sse < best["sse"]):
                    sst = float(np.sum((y - y.mean()) ** 2))
                    r2 = 1 - sse / sst if sst > 0 else np.nan
                    best = dict(success=True, B=float(beta[1]), m=float(m), omega=float(omega),
                                tc=float(tc), r2=r2, sse=sse, n=n)
    return best if best is not None else {"success": False, "n": len(y)}

def classify_fit(fit: dict):
    if not fit.get("success", False):
        return False, "NONE"
    n = fit["n"]
    m, omega, tc, r2 = fit["m"], fit["omega"], fit["tc"], fit["r2"]
    valid = True
    if not (0.1 <= m <= 0.9): valid = False
    if not (6.0 <= omega <= 13.0): valid = False
    if not ((n-1) < tc <= (n-1) + 1.5*n): valid = False
    if np.isnan(r2) or r2 < 0.2: valid = False
    if not valid:
        return False, "NONE"
    # Convention : B < 0 => "UP", B > 0 => "DOWN"
    return True, ("UP" if fit.get("B", np.nan) < 0 else "DOWN")

def compute_details(weekly_slice: pd.DataFrame) -> pd.DataFrame:
    rows = []
    for asset in weekly_slice.columns:
        s = weekly_slice[asset].dropna()
        if len(s) < max(WINDOW_WEEKS):
            continue
        vals = s.values.astype(float)
        for w in WINDOW_WEEKS:
            y = np.log(vals[-w:])
            fit = fast_lppls_fit_tiny(y)
            is_valid, signal = classify_fit(fit)
            tc_weeks_ahead = (fit["tc"] - (fit["n"] - 1)) if fit.get("success", False) else np.nan
            rows.append({
                "asset": asset,
                "window_weeks": w,
                "is_valid": bool(is_valid),
                "signal": signal,
                "tc_weeks_ahead": tc_weeks_ahead,
                "r2": fit.get("r2", np.nan),
            })
    return pd.DataFrame(rows)

def make_wide(details: pd.DataFrame) -> pd.DataFrame:
    def _wide(metric: str):
        tmp = details.pivot(index="asset", columns="window_weeks", values=metric)
        tmp.columns = [f"{metric}_{c}w" for c in tmp.columns]
        return tmp
    wide = pd.concat([_wide("signal"), _wide("is_valid"), _wide("tc_weeks_ahead"), _wide("r2")], axis=1).reset_index()
    valid_cols = [c for c in wide.columns if c.startswith("is_valid_")]
    wide["valid_frac"] = wide[valid_cols].mean(axis=1)
    tc_vals = []
    for _, row in wide.iterrows():
        vals = []
        for w in WINDOW_WEEKS:
            if row.get(f"is_valid_{w}w", False) and pd.notna(row.get(f"tc_weeks_ahead_{w}w", np.nan)):
                vals.append(row[f"tc_weeks_ahead_{w}w"])
        tc_vals.append(np.median(vals) if len(vals) > 0 else np.nan)
    wide["tc_weeks_ahead_median"] = tc_vals
    return wide

def compute_bubble_type(wide: pd.DataFrame) -> pd.DataFrame:
    short_ws = [13, 26]
    long_ws = [78, 104]

    def agg_signal(row, ws):
        sigs = []
        for w in ws:
            s = row.get(f"signal_{w}w", None)
            v = row.get(f"is_valid_{w}w", False)
            if bool(v) and isinstance(s, str) and s in ("UP", "DOWN"):
                sigs.append(s)
        if len(sigs) == 0:
            return "NONE"
        up = sum(1 for x in sigs if x == "UP")
        down = sum(1 for x in sigs if x == "DOWN")
        if up > down:
            return "UP"
        if down > up:
            return "DOWN"
        return "MIXED"

    rows = []
    for _, row in wide.iterrows():
        st = agg_signal(row, short_ws)
        lt = agg_signal(row, long_ws)
        if st in ("UP", "DOWN") and lt == st:
            mult = f"COHERENT_MULTI_{st}"
        elif st == "NONE" and lt == "NONE":
            mult = "NO_BUBBLE"
        else:
            mult = "MIXED_SCALES"

        tc_med = row["tc_weeks_ahead_median"]
        horizon_penalty = 1.0 if np.isnan(tc_med) else (1.0 + max(tc_med, 0) / 10.0)
        coh_bonus = 2.0 if mult.startswith("COHERENT_MULTI_") else (0.5 if mult == "MIXED_SCALES" else 0.0)
        priority = 100.0 * (coh_bonus + row["valid_frac"]) / horizon_penalty

        rows.append({"asset": row["asset"], "bubble_multiscale_state": mult, "priority_score": priority})
    return pd.DataFrame(rows)

def build_targets(weekly_slice: pd.DataFrame, bubble_type: pd.DataFrame) -> pd.DataFrame:
    rets = weekly_slice.pct_change(fill_method=None)
    vol_52 = rets.tail(52).std(skipna=True)
    inv_vol = (1 / vol_52.replace(0, np.nan)).replace([np.inf, -np.inf], np.nan)

    bt = bubble_type.set_index("asset")

    # Long-only : on évite COHERENT_MULTI_UP
    eligible_long = bt[bt["bubble_multiscale_state"] != "COHERENT_MULTI_UP"].index
    w_long = inv_vol.reindex(eligible_long).fillna(inv_vol.median()).clip(lower=0)
    w_long = w_long / w_long.sum() if w_long.sum() > 0 else w_long

    # Long/Short : on short COHERENT_MULTI_UP (pondéré par priority_score), long le reste en inv-vol
    short_assets = bt[bt["bubble_multiscale_state"] == "COHERENT_MULTI_UP"].index
    w_short = bt.loc[short_assets, "priority_score"].clip(lower=0)
    w_short = (w_short / w_short.sum()) if (len(w_short) > 0 and w_short.sum() > 0) else pd.Series(dtype=float)

    w_long_ls = inv_vol.reindex(eligible_long).fillna(inv_vol.median()).clip(lower=0)
    w_long_ls = (w_long_ls / w_long_ls.sum()) if w_long_ls.sum() > 0 else w_long_ls

    gross_long = 1.0 if len(w_short) == 0 else 0.5
    gross_short = 0.0 if len(w_short) == 0 else 0.5

    out = pd.DataFrame(index=bt.index)
    out["state"] = bt["bubble_multiscale_state"]
    out["priority_score"] = bt["priority_score"]
    out["w_lo_target"] = out.index.map(w_long).fillna(0.0)
    out["w_ls_target"] = out.index.map(w_long_ls).fillna(0.0) * gross_long - out.index.map(w_short).fillna(0.0) * gross_short
    return out.reset_index().rename(columns={"index": "asset"})

# =========================
# Boucle backtest
# =========================
prev_state = pd.Series(index=assets, dtype=object)
prev_w_lo = pd.Series(0.0, index=assets)
prev_w_ls = pd.Series(0.0, index=assets)

portfolios_lo = []
portfolios_ls = []
turnover = []
states = []

for k, t in enumerate(rebalance_dates):
    up_to_t = weekly.loc[:t]
    lookback = up_to_t.tail(lookback_max)

    pos = weekly.index.get_loc(t)
    forward_dates = weekly.index[pos + 1: pos + 1 + hold_weeks]

    # Éligible = historique suffisant + données présentes sur l'horizon de détention
    eligible = (lookback.notna().sum(axis=0) >= lookback_max) & (weekly.loc[forward_dates].notna().sum(axis=0) >= hold_weeks)
    eligible_assets = weekly.columns[eligible].tolist()

    details = compute_details(up_to_t[eligible_assets])
    wide = make_wide(details)
    bt = compute_bubble_type(wide)
    targets = build_targets(up_to_t[eligible_assets], bt)

    cur_state = pd.Series("INELIGIBLE", index=assets, dtype=object)
    cur_state.loc[eligible_assets] = targets.set_index("asset")["state"]

    target_lo = pd.Series(0.0, index=assets)
    target_ls = pd.Series(0.0, index=assets)
    tt = targets.set_index("asset")
    target_lo.loc[eligible_assets] = tt["w_lo_target"]
    target_ls.loc[eligible_assets] = tt["w_ls_target"]

    # Lazy rebalance : on ne touche que les actifs dont le statut change
    if k == 0:
        w_lo = target_lo.copy()
        w_ls = target_ls.copy()
    else:
        changed = (cur_state != prev_state)
        w_lo = prev_w_lo.copy()
        w_ls = prev_w_ls.copy()
        w_lo.loc[changed] = target_lo.loc[changed]
        w_ls.loc[changed] = target_ls.loc[changed]

    portfolios_lo.append(pd.DataFrame({"date": t.date().isoformat(), "asset": assets, "weight": w_lo.values}))
    portfolios_ls.append(pd.DataFrame({"date": t.date().isoformat(), "asset": assets, "weight": w_ls.values}))

    to_lo = 0.5 * np.sum(np.abs(w_lo.values - (0.0 if k == 0 else prev_w_lo.values)))
    to_ls = 0.5 * np.sum(np.abs(w_ls.values - (0.0 if k == 0 else prev_w_ls.values)))
    turnover.append({"date": t.date().isoformat(), "turnover_long_only": float(to_lo), "turnover_long_short": float(to_ls)})

    states.append(pd.DataFrame({"date": t.date().isoformat(), "asset": assets, "state": cur_state.values}))

    prev_state = cur_state
    prev_w_lo = w_lo
    prev_w_ls = w_ls

weights_lo = pd.concat(portfolios_lo, ignore_index=True)
weights_ls = pd.concat(portfolios_ls, ignore_index=True)
turnover_df = pd.DataFrame(turnover)
states_df = pd.concat(states, ignore_index=True)

# =========================
# Sauvegarde : format long + wide
# =========================
weights_lo.to_csv(OUTDIR / "backtest_portfolios_long_only_weights.csv", index=False)
weights_ls.to_csv(OUTDIR / "backtest_portfolios_long_short_weights.csv", index=False)
turnover_df.to_csv(OUTDIR / "backtest_turnover.csv", index=False)
states_df.to_csv(OUTDIR / "backtest_states.csv", index=False)

# Wide (dates x actions)
weights_lo_wide = weights_lo.pivot(index="date", columns="asset", values="weight").fillna(0.0)
weights_ls_wide = weights_ls.pivot(index="date", columns="asset", values="weight").fillna(0.0)
weights_lo_wide.to_csv(OUTDIR / "backtest_portfolios_long_only_weights_wide.csv")
weights_ls_wide.to_csv(OUTDIR / "backtest_portfolios_long_short_weights_wide.csv")

# =========================
# NAV hebdomadaire
# =========================
wlo_by_date = {d: g.set_index("asset")["weight"] for d, g in weights_lo.groupby("date")}
wls_by_date = {d: g.set_index("asset")["weight"] for d, g in weights_ls.groupby("date")}
reb_dates = sorted([pd.Timestamp(d) for d in wlo_by_date.keys()])

nav_lo = 1.0
nav_ls = 1.0
rows = []
active_reb = reb_dates[0]
weeks_bt = weekly.index[weekly.index >= active_reb]

for i, tt in enumerate(weeks_bt):
    if tt in reb_dates:
        active_reb = tt
    if i == 0:
        rows.append({"date": tt.date().isoformat(), "nav_long_only": nav_lo, "nav_long_short": nav_ls,
                     "ret_long_only": 0.0, "ret_long_short": 0.0,
                     "active_rebalance_date": active_reb.date().isoformat()})
        continue

    w_lo = wlo_by_date[active_reb.date().isoformat()].reindex(assets).fillna(0.0)
    w_ls = wls_by_date[active_reb.date().isoformat()].reindex(assets).fillna(0.0)
    r = weekly_rets.loc[tt].reindex(assets).fillna(0.0)

    ret_lo = float((w_lo * r).sum())
    ret_ls = float((w_ls * r).sum())
    nav_lo *= (1 + ret_lo)
    nav_ls *= (1 + ret_ls)

    rows.append({"date": tt.date().isoformat(), "nav_long_only": nav_lo, "nav_long_short": nav_ls,
                 "ret_long_only": ret_lo, "ret_long_short": ret_ls,
                 "active_rebalance_date": active_reb.date().isoformat()})

nav = pd.DataFrame(rows)
nav.to_csv(OUTDIR / "backtest_nav_weekly.csv", index=False)

# =========================
# Métriques
# =========================
def max_drawdown(nav_series: pd.Series) -> float:
    cm = nav_series.cummax()
    dd = 1 - nav_series / cm
    return float(dd.max())

def sharpe_ratio(returns: pd.Series, periods_per_year=52) -> float:
    mu = returns.mean()
    sd = returns.std(ddof=1)
    if sd == 0 or np.isnan(sd):
        return np.nan
    return float((mu / sd) * np.sqrt(periods_per_year))

metrics = pd.DataFrame([{
    "start_rebalance": reb_dates[0].date().isoformat(),
    "end_week": pd.to_datetime(nav["date"]).max().date().isoformat(),
    "num_rebalances": int(len(reb_dates)),
    "turnover_mean_long_only": float(turnover_df["turnover_long_only"].mean()),
    "turnover_mean_long_short": float(turnover_df["turnover_long_short"].mean()),
    "max_drawdown_long_only": max_drawdown(nav["nav_long_only"]),
    "max_drawdown_long_short": max_drawdown(nav["nav_long_short"]),
    "sharpe_long_only_weekly": sharpe_ratio(nav["ret_long_only"].iloc[1:]),
    "sharpe_long_short_weekly": sharpe_ratio(nav["ret_long_short"].iloc[1:]),
}])
metrics.to_csv(OUTDIR / "backtest_summary_metrics.csv", index=False)

# =========================
# Graphe NAV
# =========================
plt.figure(figsize=(10, 5))
plt.plot(pd.to_datetime(nav["date"]), nav["nav_long_only"], label="Long-only")
plt.plot(pd.to_datetime(nav["date"]), nav["nav_long_short"], label="Long/Short/Flat")
plt.legend()
plt.title("Backtest NAV (hebdomadaire) – holding 4 semaines")
plt.xlabel("Date")
plt.ylabel("NAV")
plt.tight_layout()
plt.savefig(OUTDIR / "backtest_nav_plot.png", dpi=160)
plt.close()

print("OK - livrables générés dans :", OUTDIR.resolve())
```
