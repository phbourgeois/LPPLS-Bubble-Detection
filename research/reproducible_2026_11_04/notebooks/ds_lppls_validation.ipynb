## Configuration

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / 'lppls_core.py').exists() and (ROOT.parent / 'lppls_core.py').exists():
    ROOT = ROOT.parent
sys.path.insert(0,str(ROOT))
from lppls_core import load_prices, to_weekly, write_snapshot_outputs, run_backtest
INPUT_CSV = ROOT / 'Miscellanous.csv'  # modifier pour utiliser un autre CSV
OUTPUT_DIR = ROOT / 'outputs'
OUTPUT_DIR.mkdir(parents=True,exist_ok=True)
print('Source:',INPUT_CSV,'| Sortie:',OUTPUT_DIR)
```

## Exécution / contrôle

```python
from pathlib import Path
assert (ROOT/'lppls_core.py').exists()
expected=['details_by_scale_all_series.csv','summary_by_scale_wide.csv',
'bubble_type_by_asset.csv','bubble_summary_by_scale_wide.csv',
'portfolio_construction.csv','heatmap_global.png']
missing=[x for x in expected if not (OUTPUT_DIR/x).exists()]
if missing: print('Exécuter ds_lppls_pipeline.ipynb d’abord. Manquants:',missing)
else:
    import pandas as pd
    a=pd.read_csv(OUTPUT_DIR/expected[0]); b=pd.read_csv(OUTPUT_DIR/expected[2]);p=pd.read_csv(OUTPUT_DIR/expected[4])
    assert len(a)>0 and len(b)>0 and len(p)>0
    assert set(b.asset)==set(p.asset)
    assert set(a.signal.unique()) <= {'UP','DOWN','NONE'}
    print('Contrôles de forme des livrables: OK')
```
